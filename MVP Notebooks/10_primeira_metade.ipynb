# Objetivo



Para este projeto, optei por utilizar a mesma base de dados que explorei no meu MVP do sprint anterior, focado em Machine Learning. Atualmente, atuo na área comercial B2B de uma grande empresa de telecomunicações. Embora este *dataset* (IBM Telco Customer Churn) retrate um cenário de clientes pessoa física, ele traz variáveis e permite análises de problemas que fazem parte do meu dia a dia profissional na gestão comercial.
## Problema a ser resolvido

No setor de telecomunicações, a retenção de clientes (combate ao *churn*) e a maximização da receita sobre os contratos ativos são desafios constantes. O objetivo deste MVP é construir um pipeline de dados analítico (arquitetura *Data Lakehouse*) que processe o histórico de assinaturas, detalhando os pacotes de conectividade e os modelos de contratação.



Com a inclusão de novas bases demográficas e geográficas neste sprint, busca-se entender quais características dos serviços, dos contratos e da localização mais influenciam a fidelidade do cliente e o faturamento. Para viabilizar a arquitetura no Databricks, também foi incorporada uma etapa técnica de normalização de *schema* logo no início do projeto (convertendo espaços em colunas, como `Zip Code` para `Zip-Code` e `Customer ID` para `customerID`), adequando os dados às exigências do formato Delta Lake e garantindo, desde a Bronze, um identificador de cliente consistente entre todas as fontes. Esse pipeline permitirá, na prática, direcionar esforços comerciais e propostas de renovação de forma mais estratégica.
